"Gráfico barra relativo a variavel stabf, com valores e percentuais da distribuição"

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Configurando o estilo do gráfico
sns.set_theme(style="whitegrid")
plt.figure(figsize=(8, 6))

# Calculando contagens e percentuais
counts = df['stabf'].value_counts()
percentages = df['stabf'].value_counts(normalize=True) * 100

# Criando o gráfico de barras
ax = sns.barplot(x=counts.index, y=counts.values, palette="viridis")

# Adicionando os valores e percentuais sobre as barras
for i, col in enumerate(counts.index):
    count = counts[col]
    pct = percentages[col]
    ax.annotate(f'{count}\n({pct:.2f}%)',
                xy=(i, count),
                xytext=(0, 5),
                textcoords="offset points",
                ha='center', va='bottom', fontsize=11, fontweight='bold')

# Ajustando títulos e rótulos
plt.title('Distribuição da Variável stabf (Valores e Percentuais)', fontsize=14, pad=20)
plt.xlabel('stabf', fontsize=12)
plt.ylabel('Frequência Absoluta', fontsize=12)
plt.ylim(0, max(counts.values) * 1.15)  # Espaço extra no topo para as labels

plt.show()
```
